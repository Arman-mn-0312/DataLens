```python
missing_percentage = (df.isnull().sum() / df.shape[0]) * 100      #we can also write len(df)in place of df.shape[0]
missing_percentage
```

```python
# Calculate missing percentage
missing_percentage = (df.isnull().sum() / len(df)) * 100

# Empty list to store report
report = []

# Generate report
for column in missing_percentage.index:            # .index will provide cloumns name

    percentage = missing_percentage[column]

    if percentage == 0:
        severity = "No Issue"

    elif percentage <= 5:
        severity = "Low"

    elif percentage <= 20:
        severity = "Medium"

    else:
        severity = "High"

    record = {
        "Column": column,
        "Missing Percentage": round(percentage, 2),
        "Severity": severity
    }

    report.append(record)

print(report)
```

```python
report_df = pd.DataFrame(report)
report_df
```

# Dataset Overview Report

```python
overview = {

    "Rows": df.shape[0],

    "Columns": df.shape[1],

    "Numeric Columns":
    len(df.select_dtypes(include=["number"]).columns),

    "Categorical Columns":
    len(df.select_dtypes(include=["object", "category"]).columns),

    "Datetime Columns":
    len(df.select_dtypes(include=["datetime"]).columns),

    "Memory Usage (MB)":
    round(
        df.memory_usage(deep=True).sum() / (1024 * 1024),
        2
    ),

    "Missing Values":
    df.isnull().sum().sum(),

    "Duplicate Rows":
    df.duplicated().sum()
}

overview_report_df = pd.DataFrame(
    overview.items(),
    columns=["Metric", "Value"]
)

overview_report_df
```

# Missing Value Report

```python
report = []

for column in df.columns:

    missing_count = df[column].isnull().sum()

    missing_percentage = (
        missing_count / len(df)
    ) * 100

    if missing_percentage == 0:
        severity = "None"
        recommendation = "No action needed"

    elif missing_percentage <= 5:
        severity = "Low"
        recommendation = "Fill missing values"

    elif missing_percentage <= 30:
        severity = "Medium"
        recommendation = "Investigate and impute"

    else:
        severity = "High"
        recommendation = "Review column usefulness"

    report.append({

        "Column": column,

        "Missing Count": missing_count,

        "Missing %": round(
            missing_percentage,
            2
        ),

        "Severity": severity,

        "Recommendation": recommendation

    })

missing_report_df = pd.DataFrame(report)

missing_report_df
```

# Duplicate Detection Report

```python
total_rows = len(df)

duplicate_count = df.duplicated().sum()

duplicate_percentage = (
    duplicate_count / total_rows
) * 100

if duplicate_count == 0:
    status = "Clean"
    recommendation = "No action needed"

else:
    status = "Review Required"
    recommendation = (
        "Review duplicate records before removing"
    )

report = {

    "Total Rows": total_rows,

    "Duplicate Rows": duplicate_count,

    "Duplicate %": round(
        duplicate_percentage,
        2
    ),

    "Status": status,

    "Recommendation": recommendation

}

duplicate_report_df = pd.DataFrame(
    report.items(),
    columns=["Metric", "Value"]
)

duplicate_report_df
```

# Data type Report 

```python
report = []

for column in df.columns:
    current_type = str(df[column].dtype)

    if current_type == "object":
        recommended = "category"
        status = "Recommendation"
    else:
        recommended = current_type
        status = "Correct"

    record = {
        "Column" : column,
        "Current Type" : current_type,
        "Recommended Type" : recommended,
        "Status" : status
    }

    report.append(record)

datatype_report_df = pd.DataFrame(report)
datatype_report_df
```

# Outliers Report

```python
outlier_report = []

for column in df.select_dtypes(include=["number"]).columns:

    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - (1.5 * iqr)
    upper_bound = q3 + (1.5 * iqr)

    outliers = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    )

    outlier_count = outliers.sum()

    status = (
        "Clean"
        if outlier_count == 0
        else "Review Required"
    )

    outlier_report.append({
                "Column": column,
                "Q1": q1,
                "Q3": q3,
                "IQR": iqr,
                "Lower Bound": lower_bound,
                "Upper Bound": upper_bound,
                "Outlier Count": outlier_count,
                "Status": status
            })

outlier_report_df = pd.DataFrame(outlier_report)

outlier_report_df
```

# Dashboard

```python
dashboard = {
    "Total Rows": df.shape[0],

    "Total Columns": df.shape[1],

    "Columns with Missing Values":
    (missing_report_df["Missing Count"] > 0).sum(),

    "Duplicate Rows":
    df.duplicated().sum(),

    "Data Type Issues":
    (datatype_report_df["Status"] == "Recommendation").sum(),

    "Outlier Columns":
    (outlier_report_df["Outlier Count"] > 0).sum()
}

dashboard_report_df = pd.DataFrame(
    dashboard.items(),
    columns=["Metric","Value"]
)

dashboard_report_df
```
